#4. MODELAGEM FUNCIONAL
**Entrada:**
* Imagens diversas que necessitam de análise detalhada, como fotos de acidentes, superfícies danificadas, exames médicos, imagens industriais, objetos com falhas estruturais, ou qualquer outro contexto onde a segmentação e melhoria de contraste sejam úteis.

**Processamento:**
1. Filtragem de Imagens: Remoção de ruídos e melhoria da nitidez para facilitar a análise visual.
2. Processamento de Cores: Identificação de variações de tonalidade para destacar padrões e diferenças relevantes (exemplo: queimaduras, rachaduras, texturas específicas).
3. Histograma e Equalização: Ajuste de contraste e brilho para revelar detalhes ocultos em áreas muito escuras ou muito claras.
4. Morfologia Matemática: Segmentação e separação de partes de interesse na imagem (exemplo: destacar bordas, remover pequenos detalhes irrelevantes).
5. Segmentação Avançada: Aplicação de pelo menos uma das técnicas a seguir para refinar a separação de objetos e regiões importantes:
* Labeling: Identificação e contagem de objetos distintos na imagem.
* Watershed: Separação de elementos sobrepostos na cena.
* Transformada de Distância: Detecção de proximidade entre regiões de interesse.

**Saída:**
* Imagem processada com realce de áreas de interesse, segmentação clara dos elementos principais e aprimoramento de contraste e detalhes.

Cada bloco do sistema será detalhado indicando:
* Entrada: Tipo de imagem analisada e seus possíveis problemas (baixo contraste, objetos sobrepostos, ruído, etc.).
* Processamento: Técnicas aplicadas e justificativa do uso de cada uma.
* Saída: Resultado esperado após o aprimoramento e segmentação.
